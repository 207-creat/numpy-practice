```python
#单位矩阵
import numpy as np
arr=np.eye(3,4,dtype=int)
print(arr)
```

```python
#主对角线元素不为0的数值
arr=np.diag([1,5,2,3])
print(arr)
```

```python
# 随机生成0-1的浮点数
arr=np.random.rand(3,4)
print(arr)
```

```python
#生成指定范围区间的随机浮点数
arr=np.random.uniform(3,6,(2,3))
print(arr)
```

```python
#生成指定范围的随机整数
arr=np.random.randint(3,30,(3,4))
print(arr)
```

```python
#生成随机数列（正态分布）
arr=np.random.randn(3,4)
print(arr)
```

```python
#设置随机种子
np.random.seed(10)
arr=np.random.randint(1,20,(2,3))
print(arr)
```

```python
arr=np.array([1,0,5,0],dtype='bool')
print(arr)
```

```python
arr=np.array([1,0,5,128],dtype='int16')
print(arr)
```

```python
#等差数列（start，stop，step）
arr=np.arange(1,10,1)
print(arr)
```

```python
#（start，stop，num）共生成num个元素
arr=np.linspace(1,10,5,dtype=int)
print(arr)
```

```python
#一维数组索引与切片
arr=np.random.randint(1,100,20)
print(arr)
```

```python
print(arr[0])
print(arr[:])
print(arr[2:7])#左包右不包
print(arr[slice(2,7)])
print(arr[slice(2,7,3)])
print(arr[(arr>10)&(arr<70)])#布尔索引
```

```python
arr=np.random.randint(1,100,(4,8))
print(arr)
```

```python
print(arr[1,2])
print(arr[2,2:5])
print(arr[arr<50])
print(arr[2][arr[2]>50])
```

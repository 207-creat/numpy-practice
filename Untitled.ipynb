```python
import numpy as np
```

```python
arr=np.array(1)
print(arr)
```

```python
arr=np.array([1,2.5,3])
print(arr)
print("数组的形状：",arr.shape)
print("数组的维度：",arr.ndim)
print("数组的数量：",arr.size)
print("元素的类型:",arr.dtype)
```

```python
arr=np.array([[1,2,3],[4,5,6]])
print(arr)
print("数组的形状：",arr.shape)
print("数组的维度：",arr.ndim)
print("数组的数量：",arr.size)
print("元素的类型:",arr.dtype)
```

```python
list=[1,2,3]
arr=np.array(list)
print(arr.ndim)
print(arr)
```

```python
arr=np.zeros([3,4],dtype=float)
print(arr)
```

```python
arr=np.ones([3,4],dtype=int)
print(arr)
```

```python
arr=np.full([3,4],2026)
print(arr)
```

```python
arr1=np.copy(arr)
print(arr1)
arr1[0]=6
print(arr1)
```

```python
arr=np.empty((2,3))
print(arr)
```

```python
arr=np.empty((4,3))
print(arr)
```

```python
arr=np.full([5,4],2026)
print(arr)
```

```python
arr1=np.zeros_like(arr)
print(arr1)
arr1=np.empty_like(arr)
print(arr1)
arr1=np.ones_like(arr)
print(arr1)
```

```python
arr=np.arange(0,11,2)
print(arr)
```

```python
arr=np.linspace(1,10,3,dtype=float)
print(arr)
```

```python
arr=np.logspace(0,4,3,base=2)
print(arr)
```

```python
arr=np.logspace(0,10,5,base=2)
print(arr)
```
